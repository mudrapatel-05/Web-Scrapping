#### 2. Use Selenium to navigate to the Zomato website, wait for the search bar to load, and extract the placeholder text from the search input field.<br><br><em><strong>Hint:</strong> Use WebDriverWait with expected_conditions to wait for the element to be present before extracting its attribute.</em>

```python
# pip install selenium

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

driver = webdriver.Chrome()

driver.get("https://www.zomato.com")

wait = WebDriverWait(driver, 10)

search_box = wait.until(
    EC.presence_of_element_located(
        (By.CSS_SELECTOR, "input")
    )
)

print("Search Bar Placeholder:", search_box.get_attribute("placeholder"))

driver.quit()
```

#### 3. Write a Selenium script to open the BookMyShow homepage, scroll down to load more content dynamically, and print the names of the first 5 movies displayed.<br><br><em><strong>Hint:</strong> Use driver.execute_script('window.scrollTo(...)') to scroll, and find_elements to get movie names.</em>

```python
# pip install selenium

from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()

driver.get("https://in.bookmyshow.com/")

time.sleep(5)

# Scroll down
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

time.sleep(3)

# Find movie names
movies = driver.find_elements(By.CSS_SELECTOR, "div")

count = 0

for movie in movies:
    text = movie.text.strip()

    if text and len(text) > 2:
        print(text)
        count += 1

    if count == 5:
        break

driver.quit()
```

#### 5. Use ChatGPT or another AI tool to generate a Selenium script that logs into a demo site with a username and password, waits for a welcome message, and prints it. Then, run the script and paste the code you used along with a screenshot of the output.

```python
# pip install selenium

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

driver = webdriver.Chrome()

driver.get("https://the-internet.herokuapp.com/login")

wait = WebDriverWait(driver, 10)

username = wait.until(
    EC.presence_of_element_located((By.ID, "username"))
)

password = driver.find_element(By.ID, "password")

username.send_keys("tomsmith")
password.send_keys("SuperSecretPassword!")

login_button = driver.find_element(By.CSS_SELECTOR, "button[type='submit']")
login_button.click()

welcome_message = wait.until(
    EC.presence_of_element_located((By.ID, "flash"))
)

print("Welcome Message:")
print(welcome_message.text)

driver.quit()
```
