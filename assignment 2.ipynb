#### 3. Parse the HTML content of https://www.myntra.com using BeautifulSoup and print the title of the page.

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.myntra.com"
headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers, timeout=20)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")
title = soup.title.text.strip() if soup.title else "Title not found"

print(title)
```

#### 4. Extract and display all the unique class names used in <div> tags from the Zomato homepage (https://www.zomato.com) using BeautifulSoup.<br><br><em><strong>Hint:</strong> Use soup.find_all('div') and access the 'class' attribute.</em>

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.zomato.com"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

classes = set()

for div in soup.find_all("div"):
    if div.get("class"):
        for class_name in div.get("class"):
            classes.add(class_name)

print("Unique class names:")

for class_name in sorted(classes):
    print(class_name)
```

#### 5. Pick any public webpage of an app you use (like BookMyShow, Paytm, or IRCTC), fetch its HTML using requests, and use BeautifulSoup to find and print all elements with an id attribute.


```python
import requests
from bs4 import BeautifulSoup

url = "https://in.bookmyshow.com/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

print("ID attributes:")

for element in soup.find_all(id=True):
    print(element.get("id"))
```
