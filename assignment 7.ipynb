#### 1. Use Selenium to open https://www.myntra.com/ in a Chrome browser, locate the search box, and enter the text 'shoes' using send_keys(), then submit the search.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

driver = webdriver.Chrome()

driver.get("https://www.myntra.com/")
driver.maximize_window()

time.sleep(3)

search_box = driver.find_element(By.CSS_SELECTOR, "input[placeholder='Search for products, brands and more']")
search_box.send_keys("shoes")
search_box.send_keys(Keys.ENTER)

time.sleep(5)

driver.quit()
```

##### 2. Write a Selenium script that clicks the 'Login' button on https://www.zomato.com/ and handles the popup by closing it after it appears.<br><br><em><strong>Hint:</strong> Wait for the popup using WebDriverWait and locate the close button before clicking.</em>

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

driver = webdriver.Chrome()

driver.get("https://www.zomato.com/")
driver.maximize_window()

wait = WebDriverWait(driver, 15)

login_button = wait.until(
    EC.element_to_be_clickable(
        (By.XPATH, "//a[contains(text(),'Log in')]")
    )
)

login_button.click()

time.sleep(2)

close_button = wait.until(
    EC.element_to_be_clickable(
        (By.XPATH, "//button[@aria-label='Close']")
    )
)

close_button.click()

print("Login popup opened and closed successfully.")

driver.quit()
```

#### 3. On https://www.flipkart.com/, use Selenium to close the initial login popup, then click on the 'Mobiles' menu item to navigate to the mobiles section.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()

driver.get("https://www.flipkart.com/")
driver.maximize_window()

time.sleep(3)

# Close initial login popup
try:
    close_button = driver.find_element(
        By.XPATH, "//button[contains(text(),'✕')]"
    )
    close_button.click()
except:
    pass

time.sleep(2)

# Click Mobiles
mobiles = driver.find_element(
    By.XPATH, "//a[contains(translate(text(),'ABCDEFGHIJKLMNOPQRSTUVWXYZ','abcdefghijklmnopqrstuvwxyz'),'mobiles')]"
)

mobiles.click()

time.sleep(5)

print("Navigated to Mobiles section.")
print("Current URL:", driver.current_url)

driver.quit()
```

#### 4. Build a Selenium script to scroll down the Instagram Explore page (https://www.instagram.com/explore/) multiple times, loading more posts each time, and print the total number of post links collected after each scroll.<br><br><em><strong>Hint:</strong> Use execute_script('window.scrollTo(0, document.body.scrollHeight)') and time.sleep() between scrolls.</em>

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()

driver.get("https://www.instagram.com/explore/")
driver.maximize_window()

time.sleep(5)

for i in range(5):

    # Scroll to bottom
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight)"
    )

    time.sleep(3)

    # Collect post links
    post_links = driver.find_elements(
        By.XPATH,
        "//a[contains(@href, '/p/') or contains(@href, '/reel/')]"
    )

    # Remove duplicate URLs
    links = set()

    for link in post_links:
        href = link.get_attribute("href")
        if href:
            links.add(href)

    print(f"After scroll {i + 1}: {len(links)} post links collected")

driver.quit()
```

#### 5. Use ChatGPT or Copilot to generate a Selenium code snippet for scraping product names from an infinite scrolling page like https://www.nykaa.com/search/result/?q=lipstick. Test and modify the code so it collects at least 30 product names and prints them.

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

driver = webdriver.Chrome()

driver.get("https://www.nykaa.com/search/result/?q=lipstick")
driver.maximize_window()

time.sleep(5)

product_names = set()

for i in range(10):

    # Find product names currently loaded
    products = driver.find_elements(
        By.CSS_SELECTOR,
        "div.css-xrzmfa"
    )

    for product in products:
        name = product.text.strip()

        if name:
            product_names.add(name)

    print(f"Scroll {i + 1}: {len(product_names)} products collected")

    if len(product_names) >= 30:
        break

    # Scroll down
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight)"
    )

    time.sleep(3)

print("\nFirst 30 Product Names:")

for i, name in enumerate(list(product_names)[:30], start=1):
    print(f"{i}. {name}")

driver.quit()
```
