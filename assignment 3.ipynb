#### 1. Use the requests library in Python to make a GET request to https://jsonplaceholder.typicode.com/posts and print the status code and first 100 characters of the response text.

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

print("Status Code:", response.status_code)
print("First 100 characters:")
print(response.text[:100])
```

#### 2. Send a GET request to https://httpbin.org/headers with a custom User-Agent header set to 'FlipkartBot/1.0' and print the User-Agent value received by the server from the response.

```python
import requests

url = "https://httpbin.org/headers"

headers = {
    "User-Agent": "FlipkartBot/1.0"
}

response = requests.get(url, headers=headers)

data = response.json()

print("User-Agent received by server:")
print(data["headers"]["User-Agent"])
```

#### 3. Simulate logging into a Zomato-style food delivery site by sending a POST request to https://httpbin.org/post with form data {'username': 'foodie', 'password': 'tasty123'} using requests. Print the JSON response to confirm your data was sent.<br><br><em><strong>Hint:</strong> Use the data parameter in requests.post().</em>

```python
import requests

url = "https://httpbin.org/post"

data = {
    "username": "foodie",
    "password": "tasty123"
}

response = requests.post(url, data=data)

print(response.json())
```

#### 4. Make a GET request to https://httpbin.org/cookies/set?sessionid=12345 using requests, then make another GET request to https://httpbin.org/cookies to check if your sessionid cookie is being sent. Print the cookies received in the second response.<br><br><em><strong>Hint:</strong> Use requests.Session() to persist cookies between requests.</em>

```python
import requests

session = requests.Session()

# First request: set the cookie
url1 = "https://httpbin.org/cookies/set?sessionid=12345"

response1 = session.get(url1)

# Second request: check the cookies
url2 = "https://httpbin.org/cookies"

response2 = session.get(url2)

print("Cookies received in the second response:")
print(response2.json())
```

#### 5. Write a function check_status_code(url) that makes a GET request to the given URL and prints a message: 'Success' if status code is 200, 'Not Found' if 404, and 'Other Error' for anything else.<br><br><em><strong>Constraint:</strong> Use if-elif-else to handle the different status codes.</em>

```python
import requests

def check_status_code(url):
    response = requests.get(url)

    if response.status_code == 200:
        print("Success")
    elif response.status_code == 404:
        print("Not Found")
    else:
        print("Other Error")


check_status_code("https://jsonplaceholder.typicode.com/posts")
check_status_code("https://jsonplaceholder.typicode.com/invalid-page")
```
