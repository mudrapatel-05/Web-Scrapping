#### 2. Use BeautifulSoup's find() method to extract and print the first product title from a sample HTML file that contains a list of Zomato restaurant cards.

```python
from bs4 import BeautifulSoup

with open("zomato_sample.html", "r", encoding="utf-8") as file:
    html = file.read()

soup = BeautifulSoup(html, "html.parser")

first_product = soup.find("h2", class_="restaurant-title")

print("First restaurant title:")
print(first_product.text)
```

#### 3. Write a script that uses find_all() to collect all image tags from the homepage of Myntra and prints the src attribute of each image.<br><br><em><strong>Hint:</strong> Use requests to get the HTML, then BeautifulSoup to parse and extract the src values.</em>

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.myntra.com"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

images = soup.find_all("img")

for image in images:
    src = image.get("src")

    if src:
        print(src)
```

#### 4. Use BeautifulSoup's select() method with a CSS selector to extract all anchor tags inside a div with class 'trending' from a given HTML snippet (simulate a 'Trending Now' section like YouTube). Print the href attribute of each link.

```python
from bs4 import BeautifulSoup

html = """
<div class="trending">
    <a href="/video1">Top 10 Movies of 2026</a>
    <a href="/video2">Best Travel Destinations</a>
    <a href="/video3">Latest Technology News</a>
    <a href="/video4">Top Gaming Moments</a>
</div>

<div class="popular">
    <a href="/video5">Popular Videos</a>
</div>
"""

soup = BeautifulSoup(html, "html.parser")

links = soup.select("div.trending a")

for link in links:
    print(link.get("href"))
```

#### 5. Extract and print all the text content of the first 5 movie titles from https://quotes.toscrape.com/ using BeautifulSoup.<br><br><em><strong>Constraint:</strong> Only print the text, not the HTML tags or attributes.</em>

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

quotes = soup.select("span.text")

for quote in quotes[:5]:
    print(quote.get_text(strip=True))
```
