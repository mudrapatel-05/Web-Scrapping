#### 2. Visit Zomato.com and check the website’s robots.txt file by going to zomato.com/robots.txt — list two rules you find there and explain what parts of the site they allow or disallow for bots.

Rule 1
User-agent: Googlebot
Disallow: /admin/

Meaning: Googlebot is instructed not to crawl URLs under /admin/, so administrative pages are excluded from Google's crawling.

Rule 2
User-agent: Googlebot
Disallow: /users/

Meaning: Googlebot is instructed not to crawl URLs under /users/.

There is also a broad rule at the end:

User-agent: *
Disallow: /

This means that, for crawlers not covered by the more specific rules above, Zomato instructs them not to crawl the site.

Note: robots.txt is a crawler-access instruction, not a security mechanism. Zomato's Terms also specifically tell users to comply with instructions contained in its robots.txt file.

#### 3. Compare the homepages of Myntra.com (static content) and Instagram.com (dynamic content) by reloading each site and observing the network requests in DevTools — describe one key difference in how the main content loads for each.

Myntra

The main HTML/document and static resources such as HTML, CSS, JavaScript and images provide much of the initial page structure. Additional API requests can subsequently provide personalized or updated content.

Myntra has historically evolved from a more static storefront toward personalized/dynamic storefronts, including user-specific homepage content.

Instagram

Instagram is much more application-driven. After the initial document and JavaScript bundles load, additional API/fetch/XHR requests retrieve information such as posts, recommendations, stories, and other user-specific content.
