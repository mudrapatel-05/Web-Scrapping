#### 1. Use pandas to load a scraped restaurant dataset (CSV or JSON) and save the first 20 rows into a new file named top20_restaurants.csv.

#### 2. Clean a scraped Flipkart products dataset using pandas by replacing all missing prices with the value 0 and saving the cleaned data as cleaned_products.xlsx.

```python
import pandas as pd

# Load Flipkart products dataset
df = pd.read_csv("flipkart_products.csv")

# Replace missing values in price column with 0
df["price"] = df["price"].fillna(0)

# Save cleaned dataset as Excel
df.to_excel("cleaned_products.xlsx", index=False)

print("Missing prices replaced with 0.")
print("Cleaned file saved as cleaned_products.xlsx")
```

#### 3. Given a JSON file of IPL cricket match stats with some missing player names, use pandas to fill missing names with 'Unknown' and export the result to a new JSON file.

```python
import pandas as pd

# Load IPL JSON file
df = pd.read_json("ipl_stats.json")

# Fill missing player names
df["player_name"] = df["player_name"].fillna("Unknown")

# Save as a new JSON file
df.to_json(
    "ipl_stats_cleaned.json",
    orient="records",
    indent=4
)

print("Missing player names replaced with 'Unknown'.")
print("Cleaned JSON file saved successfully.")
```

#### 4. Take a scraped Zomato food menu CSV file and remove all rows where the 'rating' column is missing or blank using pandas, then save the cleaned data as menu_cleaned.csv.<br><br><em><strong>Hint:</strong> Use dropna() with the subset parameter for the 'rating' column.</em>

```python
import pandas as pd

# Load Zomato menu CSV
df = pd.read_csv("zomato_menu.csv")

# Convert blank ratings to missing values
df["rating"] = df["rating"].replace(
    r"^\s*$",
    pd.NA,
    regex=True
)

# Remove rows where rating is missing
df = df.dropna(subset=["rating"])

# Save cleaned dataset
df.to_csv("menu_cleaned.csv", index=False)

print("Rows with missing/blank ratings removed.")
print("Cleaned menu saved as menu_cleaned.csv")
```

#### 5. Use pandas to combine two scraped datasets: one of Spotify playlists (CSV) and one of song details (Excel), then save the merged data as merged_playlist_songs.csv.<br><br><em><strong>Hint:</strong> Use merge() and ensure you join on the song name or ID.</em>

```python
import pandas as pd

# Load Spotify playlist dataset
playlist_df = pd.read_csv("spotify_playlists.csv")

# Load song details Excel dataset
songs_df = pd.read_excel("song_details.xlsx")

# Merge datasets using song_id
merged_df = pd.merge(
    playlist_df,
    songs_df,
    on="song_id",
    how="inner"
)

# Save merged dataset
merged_df.to_csv(
    "merged_playlist_songs.csv",
    index=False
)

print("Datasets merged successfully.")
print(merged_df)
print("Merged file saved as merged_playlist_songs.csv")
```
