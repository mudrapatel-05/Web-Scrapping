#### 1. Use fetch() in JavaScript to get a list of movies from the public API https://api.tvmaze.com/shows and print the names of the first 10 shows in the console.

```javascript
fetch("https://api.tvmaze.com/shows")
    .then(response => response.json())
    .then(data => {
        data.slice(0, 10).forEach(show => {
            console.log(show.name);
        });
    })
    .catch(error => console.error("Error:", error));
```

#### 3. Write a function getWeather(city) that fetches current weather data for a city from the MetaWeather API (https://www.metaweather.com/api/) and prints the temperature. Use 'Ahmedabad' as the test city.<br><br><em><strong>Hint:</strong> You may need to first get the WOEID for the city, then fetch weather using that ID.</em>

```javascript
async function getWeather(city) {

    try {

        // First find the city coordinates
        const geoResponse = await fetch(
            `https://geocoding-api.open-meteo.com/v1/search?name=${city}&count=1`
        );

        const geoData = await geoResponse.json();

        const latitude = geoData.results[0].latitude;
        const longitude = geoData.results[0].longitude;

        // Get current weather
        const weatherResponse = await fetch(
            `https://api.open-meteo.com/v1/forecast?latitude=${latitude}&longitude=${longitude}&current=temperature_2m`
        );

        const weatherData = await weatherResponse.json();

        console.log(
            `Temperature in ${city}: ${weatherData.current.temperature_2m}°C`
        );

    } catch (error) {
        console.error("Error:", error);
    }
}

getWeather("Ahmedabad");
```

#### 4. Find a public API that requires an API key (for example, https://newsapi.org or https://openweathermap.org). Register for a free key, and write code to fetch data from one endpoint using your key. Print one interesting field from the response.<br><br><em><strong>Constraint:</strong> Do not share your API key in public code or screenshots.</em>

```javascript
const API_KEY = "YOUR_API_KEY";

const city = "Ahmedabad";

fetch(
    `https://api.openweathermap.org/data/2.5/weather?q=${city}&appid=${API_KEY}&units=metric`
)
    .then(response => response.json())
    .then(data => {
        console.log("City:", data.name);
        console.log("Temperature:", data.main.temp, "°C");
    })
    .catch(error => {
        console.error("Error:", error);
    });
```
