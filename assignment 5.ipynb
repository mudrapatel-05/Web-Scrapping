#### 1. Identify the pagination pattern used on Flipkart's search results page by inspecting the page structure and noting how the next page URL or button changes as you move from page 1 to page 2.

Task 1 — Identify Flipkart Pagination Pattern

Flipkart uses page-number-based pagination on its search results page.

For example, the first page contains the search results, and when moving to the next page, the page number in the URL/request changes from page 1 to page 2.

Example pattern:

Page 1:
https://www.flipkart.com/search?q=shoes&page=1

Page 2:
https://www.flipkart.com/search?q=shoes&page=2

Page 3:
https://www.flipkart.com/search?q=shoes&page=3

Therefore, the pagination pattern can be identified by the changing page parameter. The value of `page` increases by 1 when moving to the next page.

Conclusion:
Flipkart uses page-number-based pagination, where the page number changes for each search-results page.

#### 2. Write a Python script using requests and BeautifulSoup to scrape the names and prices of the first 20 products from the first page of a Myntra search result.

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.myntra.com/men-tshirts"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print("Status Code:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")

products = soup.select("li.product-base")

for product in products[:20]:
    name = product.select_one("h3.product-product")
    price = product.select_one("span.product-discountedPrice")

    if name and price:
        print("Product:", name.get_text(strip=True))
        print("Price:", price.get_text(strip=True))
        print("-" * 40)
```

#### 3. Extend your script to loop through the first 3 pages of Zomato restaurant search results for 'pizza' in Ahmedabad, collecting restaurant names and ratings from each page.<br><br><em><strong>Hint:</strong> Pay attention to how the URL or request parameters change for each page.</em>

```python
import requests
from bs4 import BeautifulSoup
import time

base_url = "https://www.zomato.com/ahmedabad/restaurants/pizza"

headers = {
    "User-Agent": "Mozilla/5.0"
}

for page in range(1, 4):

    print(f"\n--- PAGE {page} ---")

    params = {
        "page": page
    }

    response = requests.get(
        base_url,
        headers=headers,
        params=params
    )

    print("Status Code:", response.status_code)

    soup = BeautifulSoup(response.text, "html.parser")

    restaurants = soup.select("div")

    for restaurant in restaurants:
        text = restaurant.get_text(" ", strip=True)

        if text:
            print(text[:200])

    time.sleep(2)
```

#### 4. Modify your multi-page scraping script to include a 2-second delay between each page request to avoid getting blocked by the server.<br><br><em><strong>Constraint:</strong> Use the time.sleep() function to implement rate limiting.</em>

```python
import requests
from bs4 import BeautifulSoup
import time

url = "https://example.com/products"

headers = {
    "User-Agent": "Mozilla/5.0"
}

for page in range(1, 4):

    print(f"\nScraping Page {page}")

    params = {
        "page": page
    }

    response = requests.get(
        url,
        params=params,
        headers=headers
    )

    print("Status Code:", response.status_code)

    soup = BeautifulSoup(response.text, "html.parser")

    products = soup.select(".product")

    for product in products:
        title = product.select_one(".title")

        if title:
            print(title.get_text(strip=True))

    # Wait 2 seconds before next request
    time.sleep(2)
```

#### 5. Find a product listing website of your choice (not Flipkart, Myntra, or Zomato) and write a script that scrapes the first 5 product titles across multiple pages, handling pagination as needed.<br><br><em><strong>Hint:</strong> Choose a site with clear 'Next' or page number navigation.</em>

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

base_url = "https://books.toscrape.com/"

url = base_url

count = 0

while count < 5:

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.select("article.product_pod h3 a")

    for book in books:

        title = book.get("title")

        print(title)

        count += 1

        if count == 5:
            break

    # Find Next button
    next_button = soup.select_one("li.next a")

    if next_button:
        url = urljoin(url, next_button.get("href"))
    else:
        break
```
